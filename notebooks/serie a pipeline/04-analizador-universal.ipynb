# Analizador universal

## Configuración

```python
# 🔥 ANALIZADOR UNIVERSAL - CUALQUIER EQUIPO/JUGADOR
EQUIPO_TARGET = "Como"           # ← CAMBIA AQUÍ
JUGADOR1 = "Nico Paz"            # ← CAMBIA AQUÍ  
JUGADOR2 = "Máximo Perrone"      # ← CAMBIA AQUÍ (opcional)

print(f"🎯 ANÁLISIS: {EQUIPO_TARGET} | {JUGADOR1} + {JUGADOR2}")
print("="*60)
```

## Función motor universal

```python
def analizar_equipo_jugadores(equipo: str, jugador1: str = None, jugador2: str = None):
    """Framework universal para CUALQUIER equipo/jugador"""
    
    # Filtrar equipo
    mask_equipo = df_players['team'].str.contains(equipo, case=False, na=False)
    equipo_players = df_players[mask_equipo]
    
    print(f"🏟️ {equipo.upper()} - PLANTILLA ({len(equipo_players['player'].unique())} jugadores)")
    
    # Plantilla ordenada
    plantilla = equipo_players[['player', 'position', 'shirt_number']].drop_duplicates()
    plantilla = plantilla.sort_values('shirt_number').reset_index(drop=True)
    print(plantilla.to_string(index=False))
    
    # Stats equipo
    print(f"\n📊 {equipo.upper()} STATS:")
    print(f"Partidos: {len(equipo_players['match'].unique())}")
    print(f"Jugadores únicos: {len(plantilla)}")
    
    # Análisis jugadores específicos
    resultados = {}
    
    if jugador1:
        mask_j1 = equipo_players['player'] == jugador1
        resultados[jugador1] = equipo_players[mask_j1]
        print(f"\n⭐ {jugador1}: {len(resultados[jugador1])} PJ | Rating: {resultados[jugador1]['rating'].mean():.2f}")
    
    if jugador2:
        mask_j2 = equipo_players['player'] == jugador2
        resultados[jugador2] = equipo_players[mask_j2]
        print(f"⭐ {jugador2}: {len(resultados[jugador2])} PJ | Rating: {resultados[jugador2]['rating'].mean():.2f}")
    
    return equipo_players, resultados

# EJECUTAR
equipo_data, jugadores_data = analizar_equipo_jugadores(EQUIPO_TARGET, JUGADOR1, JUGADOR2)
```

## Visualizaciones automáticas

```python
def graficos_universales(equipo_data, jugadores_data, equipo: str, jugador1: str, jugador2: str):
    """Gráficos automáticos para CUALQUIER análisis"""
    
    # 1. TOP 10 Rating equipo
    top10 = equipo_data.nlargest(10, 'rating')[['player', 'rating', 'goals']]
    fig1 = px.bar(top10, x='rating', y='player', color='goals',
                 title=f"🏆 TOP 10 {equipo.upper()} Rating")
    fig1.show()
    
    # 2. Comparativa jugadores (si existen)
    if jugador1 in jugadores_data and len(jugadores_data[jugador1]) > 0:
        fig2 = px.scatter(jugadores_data[jugador1], x='minutesPlayed', y='rating',
                         size='goals', title=f"🌟 {jugador1} Evolution")
        fig2.show()
    
    if jugador2 in jugadores_data and len(jugadores_data[jugador2]) > 0:
        fig3 = px.scatter(jugadores_data[jugador2], x='minutesPlayed', y='rating',
                         size='goals', title=f"🔧 {jugador2} Evolution")
        fig3.show()

# GENERAR
graficos_universales(equipo_data, jugadores_data, EQUIPO_TARGET, JUGADOR1, JUGADOR2)
```
